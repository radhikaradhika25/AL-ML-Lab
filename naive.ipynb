```python

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Step 1: Load the dataset
df = pd.read_csv("weather.csv")

# Step 2: Create the target column
# Rainfall greater than 0 = Rain; otherwise = No Rain
df["RainStatus"] = df["Rainfall"].apply(
    lambda x: "Rain" if x > 0 else "No Rain"
)

# Step 3: Separate features and target
# Remove Rainfall to prevent target leakage
X = df.drop(["Rainfall", "RainStatus"], axis=1)
y = df["RainStatus"]

# Step 4: Identify numerical and categorical columns
num_cols = X.select_dtypes(include=["number"]).columns.tolist()
cat_cols = X.select_dtypes(exclude=["number"]).columns.tolist()

# Step 5: Handle missing values and encode categories
num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

cat_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", num_pipeline, num_cols),
    ("cat", cat_pipeline, cat_cols)
], sparse_threshold=0)

# Step 6: Create the Naive Bayes classifier
model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", GaussianNB())
])

# Step 7: Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42, stratify=y
)

# Step 8: Train the classifier
model.fit(X_train, y_train)

# Step 9: Predict test data
y_pred = model.predict(X_test)

# Step 10: Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Actual values:")
print(y_test.to_numpy())

print("\nPredicted values:")
print(y_pred)

print("\nTest Accuracy:", round(accuracy * 100, 2), "%")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred, labels=["No Rain", "Rain"]))

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```
